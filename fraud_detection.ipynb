```python
# Total transactions
total = len(df)

# Total fraud transactions
fraud = (df['Class'] == 1).sum()

# Transactions with amount > 100
high_amount = (df['Amount'] > 100).sum()

# Fraud transactions with amount > 100
fraud_high_amount = (
    (df['Class'] == 1) & (df['Amount'] > 100)
).sum()

# Probabilities
p_fraud = fraud / total

p_high_given_fraud = fraud_high_amount / fraud

p_high = high_amount / total

# Bayes theorem
p_fraud_given_high = (
    p_high_given_fraud * p_fraud
) / p_high

print("P(Fraud):", p_fraud)
print("P(High Amount | Fraud):", p_high_given_fraud)
print("P(High Amount):", p_high)
print("P(Fraud | High Amount):", p_fraud_given_high)
```

```python
from scipy.stats import ttest_ind

# Separate transaction amounts
normal = df[df['Class'] == 0]['Amount']
fraud = df[df['Class'] == 1]['Amount']

# Perform Welch's independent t-test
t_stat, p_value = ttest_ind(normal, fraud, equal_var=False)

# Set significance level
alpha = 0.05

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < alpha:
    print("Result: Significant difference in transaction amounts.")
    print("Reject the Null Hypothesis (H0).")
else:
    print("Result: No significant difference in transaction amounts.")
    print("Fail to reject the Null Hypothesis (H0).")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Create transaction amount categories
df['Amount_Category'] = pd.qcut(
    df['Amount'],
    q=3,
    labels=['Low', 'Medium', 'High'],
    duplicates='drop'
)

# Create contingency table
contingency_table = pd.crosstab(
    df['Amount_Category'],
    df['Class']
)

# Perform Chi-square test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Contingency Table:")
print(contingency_table)

print("\nChi-square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

if p_value < 0.05:
    print("\nReject H0: Amount category and fraud status are associated.")
else:
    print("\nFail to reject H0: No significant association found.")
```

## ML Part

```python
# Remove columns created during statistics practice
df_ml = df.drop(
    columns=['Amount_Zscore', 'Abs_Zscore', 'Amount_Category'],
    errors='ignore'
)

# Separate features and target
X = df_ml.drop('Class', axis=1)
y = df_ml['Class']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# Create Logistic Regression pipeline
lr_model = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(
        class_weight='balanced',
        max_iter=1000,
        random_state=42
    ))
])

# Train model
lr_model.fit(X_train, y_train)

# Predict on test data
y_pred_lr = lr_model.predict(X_test)

# Evaluate model
print("LOGISTIC REGRESSION RESULTS")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced_subsample',
    random_state=42,
    n_jobs=-1
)

# Train model
rf_model.fit(X_train, y_train)

# Predict on test data
y_pred_rf = rf_model.predict(X_test)

# Evaluate model
print("RANDOM FOREST RESULTS")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    average_precision_score
)

# Calculate model performance
comparison = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest'],

    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_rf)
    ],

    'Precision': [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_rf)
    ],

    'Recall': [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_rf)
    ],

    'F1 Score': [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_rf)
    ],

    'PR-AUC': [
        average_precision_score(
            y_test, lr_model.predict_proba(X_test)[:, 1]
        ),
        average_precision_score(
            y_test, rf_model.predict_proba(X_test)[:, 1]
        )
    ]
})

print(comparison.round(4).to_string(index=False))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_recall_curve
import numpy as np

# Split training data into fitting and validation sets
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Train a temporary Random Forest model
rf_tuning = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced_subsample',
    random_state=42,
    n_jobs=-1
)

rf_tuning.fit(X_fit, y_fit)

# Get fraud probabilities on validation data
val_probabilities = rf_tuning.predict_proba(X_val)[:, 1]

# Calculate precision and recall at different thresholds
precision, recall, thresholds = precision_recall_curve(
    y_val, val_probabilities
)

# Calculate F1-score for each threshold
f1_scores = (
    2 * precision[:-1] * recall[:-1] /
    (precision[:-1] + recall[:-1] + 1e-10)
)

# Select threshold with maximum F1-score
best_index = np.argmax(f1_scores)
best_threshold = thresholds[best_index]

print("Best Threshold:", round(best_threshold, 4))
print("Validation Precision:", round(precision[best_index], 4))
print("Validation Recall:", round(recall[best_index], 4))
print("Validation F1-score:", round(f1_scores[best_index], 4))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    average_precision_score
)

# Train final model on complete training data
final_model = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced_subsample',
    random_state=42,
    n_jobs=-1
)

final_model.fit(X_train, y_train)

# Get fraud probabilities on unseen test data
y_prob = final_model.predict_proba(X_test)[:, 1]

# Apply optimized threshold
best_threshold = 0.45
y_pred_final = (y_prob >= best_threshold).astype(int)

# Evaluate final model
print("FINAL RANDOM FOREST MODEL")
print("Threshold:", best_threshold)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_final))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_final,
    target_names=['Legitimate', 'Fraud']
))

print("PR-AUC:", average_precision_score(y_test, y_prob))
```

```python
import json
import pandas as pd
from kafka import KafkaConsumer

# IDs of our 20 unseen transactions
target_ids = set(test_batch.index)

consumer = KafkaConsumer(
    'transactions',
    bootstrap_servers='localhost:9092',
    auto_offset_reset='earliest',
    group_id='unseen-test-consumer',
    value_deserializer=lambda x: json.loads(x.decode('utf-8')),
    consumer_timeout_ms=30000
)

unseen_results = []

for message in consumer:

    transaction = message.value
    transaction_id = transaction["transaction_id"]

    # Ignore previously sent transactions
    if transaction_id not in target_ids:
        continue

    input_data = pd.DataFrame(
        [transaction["features"]],
        columns=feature_names
    )

    probability = model.predict_proba(input_data)[0][1]

    prediction = int(probability >= threshold)

    unseen_results.append({
        "Transaction_ID": transaction_id,
        "Fraud_Probability": round(probability * 100, 2),
        "Prediction": prediction
    })

    print(
        f"ID: {transaction_id} | "
        f"Probability: {probability * 100:.2f}% | "
        f"Prediction: {'FRAUD' if prediction else 'LEGITIMATE'}"
    )

    if len(unseen_results) == len(target_ids):
        break

consumer.close()

unseen_results_df = pd.DataFrame(unseen_results)

print("\nTotal unseen transactions processed:", len(unseen_results_df))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

# Attach actual labels separately for evaluation
unseen_results_df["Actual"] = (
    unseen_results_df["Transaction_ID"].map(y_test)
)

print("Confusion Matrix:")
print(confusion_matrix(
    unseen_results_df["Actual"],
    unseen_results_df["Prediction"]
))

print("\nClassification Report:")
print(classification_report(
    unseen_results_df["Actual"],
    unseen_results_df["Prediction"],
    target_names=["LEGITIMATE", "FRAUD"]
))
```
